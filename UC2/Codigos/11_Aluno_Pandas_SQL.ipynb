Para essa demostração vamos criar dois DataFrames e realizar merge (união).

`````python
import pandas as pd
clientes = pd.DataFrame({

    "ID": [1,2,3],

    "Nome": [
        "Ana",
        "João",
        "Maria"
    ]
})

pedidos = pd.DataFrame({

    "ID": [1,2,3],

    "Pedido": [
        "Sushi",
        "Temaki",
        "Yakisoba"
    ]
})

resultado = pd.merge(
    clientes,
    pedidos,
    on="ID"
)

print(resultado)
`````

No Pandas, podemos combinar três tabelas utilizando a função pd.merge(), que funciona de maneira semelhante ao JOIN do SQL.  

Vamos utilizar o mesmo exemplo do sistema escolar, com as tabelas Alunos, Matriculas e Cursos.  

O código abaixo cria as três tabelas, combina os dados e exibe o resultado final.  

`````python

import pandas as pd

# 1. Criando a tabela de alunos
alunos = pd.DataFrame({
    "id_aluno": [1, 2, 3],
    "nome": ["Ana", "Carlos", "Maria"]
})

# 2. Criando a tabela de matrículas
matriculas = pd.DataFrame({
    "id_matricula": [101, 102, 103],
    "id_aluno": [1, 2, 3],
    "id_curso": [10, 20, 10]
})

# 3. Criando a tabela de cursos
cursos = pd.DataFrame({
    "id_curso": [10, 20],
    "nome_curso": ["Informática", "Administração"]
})

# 4. Primeiro merge: Alunos + Matriculas
resultado = pd.merge(
    alunos,
    matriculas,
    on="id_aluno",
    how="inner" # inner join, ou seja, apenas os registros que possuem correspondência em ambas as tabelas serão incluídos no resultado
)

# 5. Segundo merge: resultado + Cursos
resultado = pd.merge(
    resultado,
    cursos,
    on="id_curso",
    how="inner"
)

# 6. Selecionando as colunas desejadas
resultado = resultado[
    ["nome", "nome_curso", "id_matricula"]
]

# 7. Exibindo o resultado
print(resultado)
`````

Cada parâmetro possui uma função:  
|Parâmetro|	Significado|
|-----|-----|
|alunos	    |Primeira tabela|
|matriculas	    |Segunda tabela|
|on="id_aluno"	|Coluna usada para relacionar as tabelas|
|how="inner"	|Mantém apenas registros que possuem correspondência nas duas tabelas|

Na segunda operação, usamos id_curso para relacionar as matrículas aos cursos.

#### Diferença entre INNER e LEFT no Pandas 

O argumento how define quais registros serão mantidos.  

|SQL	|Pandas	|Funcionamento|
|-----|------|-----|
|INNER JOIN	|how="inner"	|Somente registros correspondentes|
|LEFT JOIN	|how="left"	    |Todos os registros da tabela à esquerda|
|RIGHT JOIN	|how="right"	|Todos os registros da tabela à direita|
|FULL OUTER JOIN	|how="outer"	|Todos os registros de ambas as tabelas|

Importante: no Pandas, merge() combina dados em memória e retorna um novo DataFrame. Ele não altera automaticamente as tabelas originais.

### Conhecendo o SQLAlchemy

O SQLAlchemy permite estabelecer uma conexão com o banco de dados. O objeto central dessa conexão é chamado Engine, criado pela função create_engine(). 

Neste tutorial, vamos utilizar o SQLite, que armazena o banco em um arquivo local.

Crie um arquivo chamado teste_banco.py ou substitua temporariamente o conteúdo de main.py pelo seguinte:

`````python

from sqlalchemy import create_engine, text

# Cria uma conexão com o banco SQLite.
# O arquivo escola.db será criado se não existir.
engine = create_engine("sqlite:///escola.db")

# Testa a conexão com o banco
with engine.connect() as conexao: # realiza a conexão com o banco de dados
    resultado = conexao.execute(
        text("SELECT 1") # executa a consulta SQL para testar a conexão
    )

    print("Conexão realizada!")
    print("Resultado do teste:", resultado.scalar())
    
    
# Encerra o uso do mecanismo de conexão
engine.dispose()    
`````

A URL sqlite:///escola.db indica um banco SQLite em arquivo relativo ao diretório de execução do programa. 


### Integrando Pandas e SQLAlchemy  
Esta é a parte principal do tutorial: vamos pegar os dados do DataFrame e armazená-los em uma tabela do banco.

Crie o arquivo main.py com o programa completo abaixo. Ele cria os dados, grava a tabela, consulta o banco e recupera os registros para um DataFrame.

`````python

import pandas as pd
from sqlalchemy import create_engine, text

# 1. Criando os dados no Pandas
alunos = pd.DataFrame({
    "id_aluno": [1, 2, 3],
    "nome": ["Ana", "Carlos", "Maria"],
    "curso": [
        "Informática",
        "Administração",
        "Informática"
    ],
    "nota": [8.5, 6.0, 9.0]
})

# 2. Criando a conexão com o banco SQLite
engine = create_engine("sqlite:///escola.db")

# 3. Gravando o DataFrame em uma tabela SQL
# if_exists="replace" recria a tabela a cada execução.
alunos.to_sql(
    "alunos",
    con=engine, # conexão com o banco de dados
    if_exists="replace", # temporariamente substitui a tabela se ela já existir
    index=False # indica que não queremos salvar o índice do DataFrame como uma coluna na tabela SQL
)

print("Dados gravados no banco!")

# 4. Consultando os dados com SQL

consulta_01 = text("""
    SELECT *
    FROM alunos
""")

consulta_02 = text("""
    SELECT nome, curso, nota
    FROM alunos
    WHERE nota >= :nota_minima
    ORDER BY nota DESC
""")

# 5. Lendo o resultado SQL para um DataFrame

with engine.connect() as conexao:
    selecao = pd.read_sql(
        consulta_01,
        conexao
        )


with engine.connect() as conexao:
    aprovados = pd.read_sql(
        consulta_02,
        conexao,
        params={"nota_minima": 7}
    )

# 6. Exibindo o resultado
print(f"\n{'='*80}")
print("Todos os alunos:")
print(selecao)

print(f"\n{'='*80}")
print("\nAlunos com nota >= 7:")
print(aprovados)

print(f"\n{'='*80}")

# Encerra o uso do mecanismo de conexão
engine.dispose()
`````

O método DataFrame.to_sql() permite gravar dados em uma tabela SQL usando uma conexão compatível, incluindo o Engine do SQLAlchemy. Já pd.read_sql() permite carregar resultados de consultas para um DataFrame.

Atenção: if_exists="replace" apaga a tabela existente e a recria antes de gravar os novos dados. Em projetos reais, avalie append ou atualizações controladas para não perder informações.
