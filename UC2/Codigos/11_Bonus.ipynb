### 1. Melhorar a visualização no terminal
O Pandas permite configurar como as tabelas são exibidas.

```python
import pandas as pd

# Criando uma tabela de exemplo
alunos = pd.DataFrame({
    "Nome": ["Ana", "Carlos", "Maria", "João", "Pedro"],
    "Curso": [
        "Informática",
        "Administração",
        "Informática",
        "Redes",
        "Administração"
    ],
    "Nota": [8.5, 6.0, 9.0, 5.5, 7.0]
})

# Exibir todas as linhas
pd.set_option("display.max_rows", None)

# Exibir todas as colunas
pd.set_option("display.max_columns", None)

# Ajustar a largura de exibição
pd.set_option("display.width", 100)

# Exibir a tabela
print(alunos)
```

Essas configurações são úteis quando você tem muitas linhas ou colunas.  
No terminal, porém, as possibilidades de formatação visual são limitadas.

### 2. Formatar os números e organizar as colunas
Você pode controlar a quantidade de casas decimais e alinhar os valores.

```python
# Exibir as notas com duas casas decimais
print(alunos.to_string(
    index=False,
    formatters={"Nota": "{:.2f}".format}
))
```

O parâmetro index=False oculta a numeração das linhas, enquanto formatters define como a nota será apresentada.

### 3. Criar uma tabela colorida com Pandas
Se você estiver utilizando o Jupyter Notebook, Google Colab ou um notebook interativo do VS Code, pode usar o recurso Styler para criar tabelas mais bonitas.

```python
# Define a aparência das notas: destaca a maior, as baixas e as demais.
def destacar_notas(notas):
    maior_nota = notas.max()
    estilos = []

    for nota in notas:
        if pd.isna(nota):
            estilos.append("") # Sem estilo para valores ausentes
        elif nota == maior_nota:
            estilos.append(
                "background-color: #fef3c7; color: #92400e; font-weight: bold;"
            ) # Destaca a maior nota com fundo amarelo e texto em negrito.
        elif nota < 6:
            estilos.append(
                "background-color: #fee2e2; color: #991b1b;"
            ) # Destaca notas abaixo da média com fundo vermelho.
        else:
            estilos.append(
                "background-color: #dcfce7; color: #166534;"
            ) # Destaca notas acima da média com fundo verde.

    return estilos


# Cria a tabela estilizada a partir do DataFrame "alunos".
tabela = (
    alunos.style
    # Exibe as notas com uma casa decimal.
    .format({"Nota": "{:.2f}"})
    # Aplica cores às notas conforme o desempenho.
    .apply(destacar_notas, subset=["Nota"])
    # Ajusta o alinhamento e o espaçamento das células.
    .set_properties(**{"text-align": "center", "padding": "10px"})  # pyright: ignore[reportArgumentType]
    # Alinha os nomes e cursos à esquerda.
    .set_properties(subset=["Nome", "Curso"], **{"text-align": "left"})
    # Define o título da tabela.
    .set_caption("Desempenho dos alunos")
    # Personaliza o título, o cabeçalho e as linhas.
    .set_table_styles([
        {
            "selector": "caption",
            "props": [
                ("caption-side", "top"),
                ("font-size", "18px"),
                ("font-weight", "bold"),
                ("color", "#1e3a8a"),
                ("padding", "12px"),
            ],
        }, # Define o estilo do cabeçalho da tabela.
        {
            "selector": "thead th",
            "props": [
                ("background-color", "#1e3a8a"),
                ("color", "white"),
                ("font-weight", "bold"),
                ("text-align", "center"),
                ("padding", "10px"),
            ], #  define a formatação do cabeçalho da tabela.  
        },
        {
            "selector": "tbody tr:nth-child(even)", 
            "props": [("background-color", "#f1f5f9")],
        }, # Define a cor de fundo das linhas pares da tabela.
        {
            "selector": "table",
            "props": [
                ("border-collapse", "collapse"), # Define o colapso das bordas da tabela.
                ("font-family", "Arial, sans-serif"), # Define a fonte da tabela.
                ("border", "1px solid #cbd5e1"), #  Define a borda da tabela.
            ],
        }, # Define a formatação geral da tabela, incluindo bordas e fonte. 
        {
            "selector": "tbody tr:hover",
            "props": [("background-color", "#e0f2fe")],
        }, # Define a cor de fundo ao passar o mouse sobre as linhas da tabela.
    ])
)

# Exibe a tabela formatada no notebook.
tabela
```

O que esse código faz?
- .style: aplica estilos à tabela sem alterar os dados originais.
- .format(): exibe as notas com duas casas decimais.
- .background_gradient(): cria um gradiente de cores conforme o valor da nota.
- .set_properties(): ajusta o alinhamento e o espaçamento.
- .set_caption(): adiciona um título à tabela.
A escala de cores ajuda a identificar rapidamente notas altas e baixas.  
O estilo aparece como tabela formatada em ambientes que suportam HTML, como notebooks; no terminal, use print().

```python
# Cria uma versão estilizada do DataFrame "alunos".
tabela = (
    alunos.style
    # Exibe as notas com duas casas decimais.
    .format({"Nota": "{:.2f}"})
    # Aplica um gradiente de cores às notas, considerando a escala de 0 a 10.
    .background_gradient(subset=["Nota"], cmap="RdYlGn", vmin=0, vmax=10)
    # Centraliza o conteúdo e adiciona espaçamento às células.
    .set_properties(**{"text-align": "center", "padding": "8px"})
    # Define o título da tabela.
    .set_caption("Desempenho dos alunos")
)

# Exibe a tabela formatada no notebook.
tabela
```
