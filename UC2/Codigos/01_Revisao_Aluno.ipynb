Exercício 1: Análise de Vendas
Enunciado:
Crie um programa que analise um conjunto de dados de vendas. O programa deve ser estruturado em funções e seguir as instruções abaixo:

Função le_dados_vendas(nome_arquivo): Esta função deve ler um arquivo CSV contendo informações de vendas com as colunas: Produto, Quantidade, Preço e retornar um DataFrame.

Função calcula_receita(df): Esta função deve receber um DataFrame e calcular a receita total de cada produto (multiplicando Quantidade por Preço). A receita deve ser adicionada como uma nova coluna chamada Receita.

Função total_vendas_por_produto(df): Esta função deve agrupar os dados por produto e calcular o total de vendas (soma das quantidades) e a receita total (soma das receitas). Retorne um novo DataFrame contendo o total de vendas e a receita por produto.

Função filtra_produtos_altas_vendas(df, limite): Esta função deve filtrar o DataFrame para retornar apenas os produtos cuja receita total é superior a um valor limite passado como argumento.

Função analise_vendas(nome_arquivo): Esta função deve orquestrar as chamadas das funções anteriores, retornando o DataFrame final com os produtos que geraram mais de R$ 500,00 em receita.

```python
# Exercicio 1
import pandas as pd

caminho_arquivo_vendas = r'F:\Cursos\Analise de dados\Senac - Analise de dados\aula11\PYTHON\UC2\DATA\vendas.csv'


# Função 1: Ler os dados de vendas

def le_dados_vendas(caminho_arquivo_vendas):

    df_vendas_dados = pd.read_csv(caminho_arquivo_vendas, sep=',', encoding='utf-8')

    colunas_vendas = ['Produto', 'Quantidade', 'Preco']
    df_colunas = df_vendas_dados[colunas_vendas].copy()

    return df_colunas


# Função 2: Calcular a receita

def calcula_receita(df_colunas):

    df_colunas['Receita'] = df_colunas['Quantidade'] * df_colunas['Preco']

    return df_colunas


# Função 3: Total de vendas por produto

def total_vendas_por_produto(df_colunas):

    soma_qtd = df_colunas.groupby('Produto')['Quantidade'].sum()
    soma_receita = df_colunas.groupby('Produto')['Receita'].sum()

    df_produto = pd.DataFrame({
        'Total_Vendas': soma_qtd,
        'Receita_Total': soma_receita
    })

    return df_produto


# Função 4: Filtrar produtos com altas vendas

def filtra_produtos_altas_vendas(df_colunas, limite):

    filtro_alta_vendas = df_colunas['Receita_Total'] > limite

    vendas_altas = df_colunas[filtro_alta_vendas]

    return vendas_altas


# Função 5: Análise completa das vendas

def analise_vendas(caminho_arquivo_vendas):

    df_colunas = le_dados_vendas(caminho_arquivo_vendas)

    df_resultado = calcula_receita(df_colunas)

    df_total = total_vendas_por_produto(df_resultado)

    df_final = filtra_produtos_altas_vendas(df_total, 500)

    return df_final


# Executando a análise
resultado_final = analise_vendas(caminho_arquivo_vendas)

print(resultado_final.to_string(
    formatters={"Receita_Total": "{:.2f}".format}
))

print(f"\nReceita total: R$ {resultado_final['Receita_Total'].sum():.2f}")
```

Exercício 2: Análise de Notas de Alunos
Enunciado:
Desenvolva um programa que analise as notas de alunos. O programa deve ser estruturado em funções e seguir as instruções abaixo:

Função le_dados_alunos(nome_arquivo): Esta função deve ler um arquivo CSV contendo informações de alunos com as colunas: Nome, Nota1, Nota2, Nota3 e retornar um DataFrame.

Função calcula_media(df): Esta função deve calcular a média das notas de cada aluno e adicionar uma nova coluna chamada Media ao DataFrame.

Função verifica_aprovacao(df): Esta função deve adicionar uma nova coluna chamada Aprovado, que deve conter True se a média do aluno for maior ou igual a 7 e False caso contrário.

Função separa_alunos_aprovados_reprovados(df): Esta função deve separar os alunos em dois DataFrames: um para aprovados e outro para reprovados, com base na coluna Aprovado.

Função analise_notas(nome_arquivo): Esta função deve orquestrar as chamadas das funções anteriores e retornar os DataFrames de alunos aprovados e reprovados.

```python
# Exercicio 2

import pandas as pd

caminho_arquivo_alunos = r'F:\Cursos\Analise de dados\Senac - Analise de dados\aula11\PYTHON\UC2\DATA\notas.csv'


# Função 1: Ler os dados dos alunos

def le_dados_alunos(caminho_arquivo_alunos):

    df_dados_alunos = pd.read_csv(caminho_arquivo_alunos, sep=',', encoding='utf-8')

    colunas_alunos = ['Nome', 'Nota1', 'Nota2', 'Nota3']
    df_alunos = df_dados_alunos[colunas_alunos].copy()

    return df_alunos

# Função calcula_media(df): .

def calcula_media(df_alunos):

    df_alunos['Media'] = (df_alunos['Nota1'] + df_alunos['Nota2'] + df_alunos['Nota3']) / 3
    return df_alunos

#Função verifica_aprovacao(df): 

def verifica_aprovacao(df_alunos):

    df_alunos = calcula_media(df_alunos)

    df_alunos['Aprovado'] = df_alunos['Media'] >= 7

    return df_alunos

#Função separa_alunos_aprovados_reprovados(df): 

def separa_alunos_aprovados_reprovados(df_alunos):

    situacao_aluno = verifica_aprovacao(df_alunos)

    df_aprovado = situacao_aluno[situacao_aluno['Aprovado'] == True]

    df_reprovado = situacao_aluno[situacao_aluno['Aprovado'] == False]

    return df_aprovado, df_reprovado


#Função analise_notas(nome_arquivo): 
        
def analise_notas(df_alunos):

    df_alunos = le_dados_alunos(caminho_arquivo_alunos)

    df_media = calcula_media(df_alunos)

    df_situacao = verifica_aprovacao(df_media)

    df_aprovados, df_reprovados = separa_alunos_aprovados_reprovados(df_situacao)

    return df_aprovados, df_reprovados

df_aprovados, df_reprovados = analise_notas(caminho_arquivo_alunos)



print("\n==============ALUNOS APROVADOS==============")
print(df_aprovados.to_string(
    index=False,
    formatters={"Media": "{:.1f}".format}
))

print("\n==============ALUNOS REPROVADOS=============")
print(df_reprovados.to_string(
    index=False,
    formatters={"Media": "{:.1f}".format}
))

```

Exercício 3: Análise de Dados de Funcionários
Enunciado:
Crie um programa que analise os dados de funcionários de uma empresa. O programa deve ser estruturado em funções e seguir as instruções abaixo:

Função le_dados_funcionarios(nome_arquivo): Esta função deve ler um arquivo CSV com as colunas: Nome, Departamento, Salário e retornar um DataFrame.

Função calcula_salario_medio(df): Esta função deve agrupar os dados por departamento e calcular o salário médio de cada um, retornando um novo DataFrame com o nome do departamento e o salário médio.

Função filtra_departamentos_alto_salario(df, limite): Esta função deve filtrar o DataFrame para retornar apenas os departamentos cuja média salarial é superior a um valor limite passado como argumento.

Função analise_funcionarios(nome_arquivo): Esta função deve orquestrar as chamadas das funções anteriores e retornar o DataFrame final com os departamentos que têm um salário médio acima de R$ 5000,00.

```python
# Exercicio 3

import pandas as pd

caminho_arquivo_funcionarios = r'F:\Cursos\Analise de dados\Senac - Analise de dados\aula11\PYTHON\UC2\DATA\funcionarios.csv'

#Função le_dados_funcionarios(nome_arquivo)

def le_dados_funcionarios(caminho_arquivo_funcionarios):
    df_dados_funcionarios = pd.read_csv(caminho_arquivo_funcionarios, sep=',', encoding='utf-8')
    
    colunas_funcionarios = ['Nome', 'Departamento', 'Salário']
    df_funcionarios = df_dados_funcionarios[colunas_funcionarios].copy()

    return df_funcionarios


#Função calcula_salario_medio(df)

def calcula_salario_medio(df_funcionarios):

    sal_medio = df_funcionarios.groupby('Departamento')['Salário'].mean()

    df_salario = pd.DataFrame(sal_medio)

    return df_salario


#Função filtra_departamentos_alto_salario(df, limite)

def filtra_departamentos_alto_salario(df_salario, limite):

    filtro_alto_salario = df_salario['Salário'] > limite
    
    salarios_altos = df_salario[filtro_alto_salario]

    return salarios_altos
    

#Função analise_funcionarios(nome_arquivo)

def analise_funcionarios(caminho_arquivo_funcionarios):

    df_funcionarios = le_dados_funcionarios(caminho_arquivo_funcionarios)
    df_salario = calcula_salario_medio(df_funcionarios)
    salarios_altos = filtra_departamentos_alto_salario(df_salario, 5000)

    return salarios_altos


resultado_final = analise_funcionarios(caminho_arquivo_funcionarios)


print("\n==============SALÁRIOS MÉDIOS ACIMA DE R$ 5.000,00=============")
print(resultado_final.to_string(
    formatters={"Salário": "{:.2f}".format}
))
```

```python
# Exercício 4


import pandas as pd

caminho_arquivo = r'F:\Cursos\Analise de dados\Senac - Analise de dados\aula11\PYTHON\UC2\DATA\vendas.csv'


# Função 1: Ler o arquivo

def le_dados_vendas(caminho_arquivo):

    df_venda = pd.read_csv(caminho_arquivo, sep=',', encoding='utf-8')

    return df_venda


# Função 2: Calcular as receitas

def calcula_receitas(df_venda):

    df_venda['Receita'] = df_venda['Quantidade'] * df_venda['Preco']

    return df_venda


# Função 3: Resumir e filtrar os resultados

def resume_filtra_resultados(df_venda):

    receita_total = df_venda.groupby('Produto')['Receita'].sum()

    df_resumo = pd.DataFrame({
        'Receita_Total': receita_total
    })

    filtro = df_resumo['Receita_Total'] > 500

    df_final = df_resumo[filtro]

    return df_final


# Função principal

def main():

    df_vendas = le_dados_vendas(caminho_arquivo)

    df_receitas = calcula_receitas(df_vendas)

    df_resultado = resume_filtra_resultados(df_receitas)

    
    print(df_resultado.to_string(
        formatters={"Receita_Total": "{:.2f}".format}
    ))




# Executar o programa

main()
```
